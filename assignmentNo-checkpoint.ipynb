```python

students = []

for i in range(2):
    name = input("Enter name: ")
    roll = input("Enter roll number: ")
    ai = int(input("Enter AI marks: "))
    python = int(input("Enter Python marks: "))
    maths = int(input("Enter Maths marks: "))

    student = {
        "name": name,
        "roll": roll,
        "ai": ai,
        "python": python,
        "maths": maths
    }

    students.append(student)


for student in students:
    obtained = student["ai"] + student["python"] + student["maths"]
    percentage = obtained / 300 * 100

    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    student["obtained"] = obtained
    student["percentage"] = percentage
    student["grade"] = grade


print("\nSTUDENT RESULTS")

for student in students:
    print("\nName:", student["name"])
    print("Roll:", student["roll"])
    print("Obtained:", student["obtained"])
    print("Percentage:", student["percentage"])
    print("Grade:", student["grade"])


highest = students[0]

for student in students:
    if student["percentage"] > highest["percentage"]:
        highest = student

print("\nHighest Percentage Student:", highest["name"])
print("Percentage:", highest["percentage"])


file = open("results.txt", "w")

for student in students:
    file.write("Name: " + student["name"] + "\n")
    file.write("Roll: " + student["roll"] + "\n")
    file.write("Obtained: " + str(student["obtained"]) + "\n")
    file.write("Percentage: " + str(student["percentage"]) + "\n")
    file.write("Grade: " + student["grade"] + "\n\n")

file.write("Highest Student: " + highest["name"] + "\n")
file.write("Percentage: " + str(highest["percentage"]) + "\n")

file.close()

print("Results saved in results.txt")
```

```python
def word(words):
    return len(words)


def sent(sentence):
    return len(sentence)


def long(words):
    longest = words[0]
    for w in words:
        if len(w) > len(longest):
            longest = w
    return longest


def short(words):
    shortest = words[0]
    for w in words:
        if len(w) < len(shortest):
            shortest = w
    return shortest


def vowel(sentence):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in sentence:
        if ch in vowels:
            count = count + 1
    return count




sentence = input("Enter a sentence: ")

words = sentence.split()

total_words = word(words)
total_characters = sent(sentence)
longest_word = long(words)
shortest_word = short(words)
total_vowels = vowel(sentence)

unique = set(words)
alphabetical = sorted(unique)

print("\nWORD ANALYSIS")

print("Sentence:", sentence)
print("Total Words:", total_words)
print("Total Characters:", total_characters)
print("Longest Word:", longest_word)
print("Shortest Word:", shortest_word)
print("Total Vowels:", total_vowels)
print("Unique Words:", unique)
print("Alphabetical Order:", alphabetical)

file = open("word_analysis.txt", "w")

file.write("WORD ANALYSIS\n")
file.write("=============\n")
file.write("Sentence: " + sentence + "\n")
file.write("Total Words: " + str(total_words) + "\n")
file.write("Total Characters: " + str(total_characters) + "\n")
file.write("Longest Word: " + longest_word + "\n")
file.write("Shortest Word: " + shortest_word + "\n")
file.write("Total Vowels: " + str(total_vowels) + "\n")
file.write("Unique Words: " + str(unique) + "\n")
file.write("Alphabetical Order: " + str(alphabetical) + "\n")

file.close()

print("\nAnalysis saved in word_analysis.txt")
```

```python

def bill(product):
    total = 0
    for p in product:
        amount = p["price"] * p["quantity"]
        total = total + amount
    return total
product = []
for i in range(5):
    p_name = input("Enter the product name: ")
    p_price = int(input("Enter the price: "))
    p_quantity = int(input("Enter the product quantity: "))
    data = {
        "name": p_name,
        "price": p_price,
        "quantity": p_quantity
    }

    product.append(data)
totel = bill(product)
print("\nShopping Bill")
for p in product:
    amount = p["price"] * p["quantity"]
    print(p["name"], "=", amount)
discount = 0
if totel > 10000:
    discount = totel * 10 / 100
final_bill = totel - discount
expensive = product[0]
for p in product:
    if p["price"] > expensive["price"]:
        expensive = p
print("Total Bill:", totel)
print("Discount:", discount)
print("Final Bill:", final_bill)
print("Most Expensive Product:", expensive["name"])
f = open("bill.txt", "w")

f.write("Shopping Bill\n")
for p in product:
    amount = p["price"] * p["quantity"]
    f.write(p["name"] + " = " + str(amount) + "\n")
f.write("Total Bill: " + str(totel) + "\n")
f.write("Discount: " + str(discount) + "\n")
f.write("Final Bill: " + str(final_bill) + "\n")
f.write("Most Expensive Product: " + expensive["name"] + "\n")
f.close()
print("Bill saved in bill.txt")
```

```python

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

def modulus(a, b):
    return a % b
while True:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    print("Select operation:")
    print("+ for Addition")
    print("- for Subtraction")
    print("* for Multiplication")
    print("/ for Division")
    print("% for Modulus")
    op = input("Enter operation: ")
    if op == "+":
        print("Result:", add(num1, num2))
    elif op == "-":
        print("Result:", subtract(num1, num2))
    elif op == "*":
        print("Result:", multiply(num1, num2))
    elif op == "/":
        if num2 != 0:
            print("Result:", divide(num1, num2))
        else:
            print("Cannot divide by zero")
    elif op == "%":
        if num2 != 0:
            print("Result:", modulus(num1, num2))
        else:
            print("Cannot use modulus with zero")
    else:
        print("Invalid operation")

    choice = input("Do you want to calculate again? (yes/no): ")

    if choice == "no":
        print("Calculator closed")
        break
```

```python

def calculate_bill(order):
    total = 0

    for item in order:
        total = total + order[item]["price"] * order[item]["quantity"]

    return total


menu = {
    1: {"name": "Burger", "price": 500},
    2: {"name": "Pizza", "price": 800},
    3: {"name": "Fries", "price": 250},
    4: {"name": "Drink", "price": 150}
}

order = {}

while True:
    print("\n===== RESTAURANT MENU =====")
    print("1. Burger  Rs. 500")
    print("2. Pizza  Rs. 800")
    print("3. Fries  Rs. 250")
    print("4. Drink  Rs. 150")
    print("5. Finish Order")

    choice = int(input("Select item: "))

    if choice == 5:
        break

    elif choice >= 1 and choice <= 4:
        quantity = int(input("Enter quantity: "))

        if choice in order:
            order[choice]["quantity"] += quantity
        else:
            order[choice] = {
                "name": menu[choice]["name"],
                "price": menu[choice]["price"],
                "quantity": quantity
            }

        print("Item added to order!")

    else:
        print("Invalid choice!")


total = calculate_bill(order)

print("\n===== YOUR BILL =====")

for item in order:
    name = order[item]["name"]
    price = order[item]["price"]
    quantity = order[item]["quantity"]
    amount = price * quantity

    print(name, "x", quantity, "=", amount)

print("Final Bill: Rs.", total)


file = open("orders.txt", "w")

file.write("===== RESTAURANT BILL =====\n")

for item in order:
    name = order[item]["name"]
    price = order[item]["price"]
    quantity = order[item]["quantity"]
    amount = price * quantity

    file.write(name + " x " + str(quantity) + " = Rs. " + str(amount) + "\n")

file.write("Final Bill: Rs. " + str(total) + "\n")

file.close()

print("Bill saved in orders.txt")
```
